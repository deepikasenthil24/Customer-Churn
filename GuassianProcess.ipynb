# Guassian Process w/ Customer Churn (Basic model)

### What is Guassian Process(GP)?


```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.gaussian_process import GaussianProcessClassifier
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score
from sklearn.model_selection import train_test_split
```

### Load, processed and prepare data

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
data = df.drop(columns = ['Row ID', 'Postal Code', 'cohort'])
```

```python
numerical_columns = data.select_dtypes(include=['number']).columns.tolist()
```

```python
data = df[numerical_columns]
data.head()
```

```python
features = data.columns.tolist()
target = 'churn'
```

```python
X = data[features]
y = data[target]
```

```python
# Normalize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#Split the data
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

### Define the Gaussian Process Model
The Radial Basis Function (RBF) kernel is commonly used

```python
# Define the kernel: Constant * RBF
kernel = C(1.0) * RBF(length_scale=1.0)
```

```python
gp_model = GaussianProcessClassifier(kernel=kernel, random_state=42)
```

```python
gp_model.fit(X_train, y_train)
```

### evalute model
We are using: cross validations in order to see our models preformance on multiple subsets of data

```python
y_pred = gp_model.predict(X_test)
```

```python
# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")
```
